```python
%matplotlib inline
from matplotlib import pyplot as plt
```

```python
# Hard-coded values: the less convenient way

values_ex_1_hardcoded = [
    0.00, 0.19, 0.36, 0.51, 0.64, 0.75, 0.84, 0.91, 0.96, 0.99, 1.00, 
    0.99, 0.96, 0.91, 0.84, 0.75, 0.64, 0.51, 0.36, 0.19, 0.00
]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_1_hardcoded, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
# List comprehension: the more convenient way
values_ex_1 = [-x**2/100 + x/5 for x in range(21)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_1, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
# Hard-coded values

values_ex_2_hard = [
    0.01, 0.01, 0.02, 0.02, 0.03, 0.04, 0.06, 0.10, 0.20, 0.50,
    1.00, 0.50, 0.20, 0.10, 0.06, 0.04, 0.03, 0.02, 0.02, 0.01, 0.01
]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_2_hard, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
# List comprehension

values_ex_2 = [0.5 / (1 + (x - 5)**2
) for x in range(30)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_2, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
values_ex_3_parametrized = [
    (x % 5) / 4 * (0.87 - 0.17 * (x // 5))
    for x in range(30)
]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_3_parametrized, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
values_ex_5 = [1 / (1 + (x - 5)**2
) for x in range(20)] + [2 / (1 + (x - 5)**2
) for x in range(20)]
peaks = (7,1),(20,2)
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_5, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
# Parametrized version
peaks = [(7, 1), (20, 2)]
values_ex_3_parametrized = [sum(A/(1 + (x - c)**2) for c, A in peaks) for x in range(30)]
plt.figure(figsize=(9, 2.5))
plt.plot(values_ex_3_parametrized, marker='o')
plt.xlabel('Sample index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```
